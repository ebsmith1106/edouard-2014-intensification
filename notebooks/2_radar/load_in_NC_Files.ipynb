```python
import xarray as xr
import os

radar_nc_dir = "NC_files"
nc_file = os.path.join(radar_nc_dir, files[0])

print("Opening file with decode_times=False")
print(nc_file)

ds = xr.open_dataset(nc_file, decode_times=False)

print("Opened successfully!")
print(ds)
```

```python
import xarray as xr
import os

radar_nc_dir = "NC_files"

radar_file = "20140911I1_BXC.nc"
radar_path = os.path.join(radar_nc_dir, radar_file)

print("Opening radar file:")
print(radar_path)

ds = xr.open_dataset(radar_path, decode_times=False)

print("\nOpened successfully!")
print(ds)
```

```python
print("\nDimensions:")
for d, n in ds.dims.items():
    print(f"{d}: {n}")
```
