```python
import requests
import os
from datetime import datetime, timedelta

# ------------------
# User settings
# ------------------
storm_dir = "2014al06"
storm_id = "2014al06"
base_url = f"https://rammb-data.cira.colostate.edu/tc_realtime/products/storms/{storm_dir}/4kmirimg"

out_dir = "edouard_4km_ir"
os.makedirs(out_dir, exist_ok=True)

start = datetime(2014, 9, 11, 14, 45)
end   = datetime(2014, 9, 20, 5, 15)
step  = timedelta(minutes=30)

# ------------------
# Download loop
# ------------------
dt = start
while dt <= end:
    timestr = dt.strftime("%Y%m%d%H%M")
    fname = f"{storm_id}_4kmirimg_{timestr}.gif"
    url = f"{base_url}/{fname}"
    out_path = os.path.join(out_dir, fname)

    r = requests.get(url)
    if r.status_code == 200:
        with open(out_path, "wb") as f:
            f.write(r.content)
        print(f"Downloaded {fname}")
    else:
        print(f"Missing {fname}")

    dt += step
```
