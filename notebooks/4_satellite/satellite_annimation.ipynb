```python
import os
import glob
from PIL import Image
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

# Directory containing the GIFs
img_dir = "edouard_4km_ir"

# Grab and sort files
files = sorted(glob.glob(os.path.join(img_dir, "*.gif")))

# Load images
images = [Image.open(f) for f in files]

# Create figure
fig, ax = plt.subplots(figsize=(6, 6))
ax.axis("off")

im = ax.imshow(images[0])

def update(frame):
    im.set_data(images[frame])
    ax.set_title(os.path.basename(files[frame]).split("_")[-1][:12])
    return [im]

ani = animation.FuncAnimation(
    fig,
    update,
    frames=len(images),
    interval=300,   # ms between frames
    blit=True
)

HTML(ani.to_jshtml())
```

```python
import os
import glob
from PIL import Image
import matplotlib.pyplot as plt
from matplotlib import animation

# Directory with downloaded GIFs
img_dir = "edouard_4km_ir"
files = sorted(glob.glob(os.path.join(img_dir, "*.gif")))

# Load images
images = [Image.open(f).convert("RGBA") for f in files]

# Create figure
fig, ax = plt.subplots(figsize=(6, 6))
ax.axis("off")

im = ax.imshow(images[0])

def update(frame):
    im.set_data(images[frame])
    return [im]

ani = animation.FuncAnimation(
    fig,
    update,
    frames=len(images),
    interval=300,   # ms between frames
    blit=True
)

# Save MP4
out_mp4 = "edouard_ir_loop.mp4"
ani.save(out_mp4, writer="ffmpeg", dpi=150)

plt.close(fig)
print(f"Saved {out_mp4}")
```
