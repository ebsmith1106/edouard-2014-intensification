```python
import pandas as pd
import re

# Read the LIS lightning text file
lis_file = "LIS_Manual_Lightning.txt"

# Lists to store data
flash_times = []
lats = []
lons = []

with open(lis_file, "r") as f:
    for line in f:
        # Match lines with a format like: ( 21.767, -45.242)
        match = re.search(r"\(\s*([0-9.+-]+),\s*([0-9.+-]+)\)", line)
        if match:
            lats.append(float(match.group(1)))
            lons.append(float(match.group(2)))
            
        # Extract flash time (UTC)
        time_match = re.match(r"(\d{4}-\d{3}T\d{2}:\d{2}:\d{2}\.\d+Z)", line)
        if time_match:
            flash_times.append(time_match.group(1))

# Create a DataFrame
flashes = pd.DataFrame({
    "flash_time": flash_times,
    "lat": lats,
    "lon": lons
})

print(flashes.head())
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from datetime import datetime

# Convert track_data to lists
track_lat = track_data["lat"]
track_lon = track_data["lon"]
track_time = [datetime.strptime(t, "%Y-%m-%d %H:%M") for t in track_data["datetime"]]

# Create figure
fig = plt.figure(figsize=(12, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

# Set map extent around the storm (adjust if needed)
ax.set_extent([-60, -40, 20, 40], crs=ccrs.PlateCarree())

# Add features
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.gridlines(draw_labels=True)

# Plot hurricane track
ax.plot(track_lon, track_lat, 'r-o', label='Hurricane Edouard Track')

# Plot lightning flashes
ax.scatter(flashes['lon'], flashes['lat'], c='yellow', s=30, alpha=0.7, edgecolors='k', label='Lightning Flashes')

plt.title("Hurricane Edouard Track with LIS Lightning Flashes")
plt.legend()
plt.show()
```
