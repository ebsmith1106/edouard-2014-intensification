```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

filename = "/nfs/home11/ugrad/2024/es348865/honors_research/TC PRIMED/AL062014/TCPRIMED_v01r01-final_AL062014_AMSR2_GCOMW1_012336_20140911152937.nc"
ds = xr.open_dataset(filename)


print(ds)  # See available variables

# Example: pick brightness temperature at 89 GHz (var names may vary slightly)
# Common var: "Tb", with dimensions (channel, y, x)
if "Tb" in ds.variables:
    tb = ds["Tb"].isel(channel=0)  # select first channel (adjust channel if needed)
else:
    raise ValueError("Tb variable not found in dataset")

# Get lat/lon
lat = ds["lat"]
lon = ds["lon"]

# Make a basic plot
plt.figure(figsize=(10, 6))
ax = plt.axes(projection=ccrs.PlateCarree())

# Plot the brightness temperatures
p = ax.pcolormesh(lon, lat, tb, cmap="turbo", shading="auto")

# Add map features
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, linestyle=":")
ax.add_feature(cfeature.STATES, linewidth=0.5)

# Colorbar
plt.colorbar(p, ax=ax, label="Brightness Temperature (K)")

ax.set_title("Edouard (2014) – Brightness Temperature Example")
plt.show()
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import pandas as pd

# = open S2 group
filename = "AL062014/TCPRIMED_v01r01-final_AL062014_AMSR2_GCOMW1_012336_20140911152937.nc"
ds = xr.open_dataset(filename, group="passive_microwave/S2")

# = pick a channel
tb = ds["TB_18.7V"]  # or TB_18.7H, etc.


scan_time = pd.to_datetime(ds["ScanTime"].values[0])


plt.figure(figsize=(10,6))
ax = plt.axes(projection=ccrs.PlateCarree())
p = ax.pcolormesh(ds["longitude"], ds["latitude"], tb, cmap="turbo", shading="auto")


ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, linestyle=":")
gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
gl.top_labels = False
gl.right_labels = False
gl.xlabel_style = {'size': 10}
gl.ylabel_style = {'size': 10}
plt.colorbar(p, ax=ax, label="Brightness Temp (K)")
ax.set_title(f"Edouard (2014) – AMSR2 S2 TB_18.7V\nScan Time: {scan_time}", fontsize=12)
plt.show()
```

```python
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# --- load dataset from S2 group ---
filename = "AL062014/TCPRIMED_v01r01-final_AL062014_AMSR2_GCOMW1_012336_20140911152937.nc"
ds = xr.open_dataset(filename, group="passive_microwave/S2")

# --- load best-track CSV ---
best_track = pd.read_csv("AL062014/AL062014_besttrack.csv")
best_track.columns = (
    best_track.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("(", "")
    .str.replace(")", "")
)
best_track = best_track[best_track['Latitude_°N'].notna()].copy()
best_track['Date/Time_UTC'] = pd.to_datetime(
    "2014-" + best_track['Date/Time_UTC'].str.replace("/", "-"),
    format="%Y-%d-%H%M",
    errors='coerce'
)

# --- pick a brightness temperature channel ---
tb = ds["TB_18.7V"]

# --- get scan date/time ---
scan_time = pd.to_datetime(ds["ScanTime"].values[0])

# --- find nearest best-track fix ---
bt_row = best_track.iloc[(best_track.Date/Time_UTC - scan_time).abs().argmin()]
storm_lat, storm_lon = bt_row["Latitude_°N"], bt_row["Longitude_°W"]

# --- plot ---
plt.figure(figsize=(10,6))
ax = plt.axes(projection=ccrs.PlateCarree())

# TB swath
p = ax.pcolormesh(ds["longitude"], ds["latitude"], tb, cmap="turbo", shading="auto")

# Coastlines and borders
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, linestyle=":")

# Lat/lon gridlines
gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.5, linestyle='--')
gl.top_labels = False
gl.right_labels = False

# Storm center
ax.plot(storm_lon, storm_lat, "r*", markersize=12, transform=ccrs.PlateCarree(), label="Best-track center")
plt.legend()

# Title with date
plt.title(f"Edouard (2014) – AMSR2 Brightness Temperature\nScan time: {scan_time}")

# Colorbar
plt.colorbar(p, ax=ax, label="Brightness Temp (K)")

plt.show()
```
