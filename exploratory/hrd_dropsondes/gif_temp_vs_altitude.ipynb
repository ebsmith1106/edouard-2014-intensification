```python
import os
from PIL import Image

# --- Path to your plots folder ---
path = "plots"

# --- Get all PNGs in sorted order ---
png_files = sorted([f for f in os.listdir(path) if f.endswith(".png")])

# --- Load images ---
frames = [Image.open(os.path.join(path, f)) for f in png_files]

# --- Save as GIF ---
output_path = os.path.join(path, "animation.gif")
frames[0].save(
    output_path,
    save_all=True,
    append_images=frames[1:],
    duration=400,   # milliseconds per frame (adjust speed here)
    loop=0          # 0 = loop forever
)

print(f"GIF saved to {output_path}")
```

```python
import os
from PIL import Image
import imageio.v2 as imageio

# --- Path to your plots folder ---
path = "plots"

# --- Get all PNGs in sorted order ---
png_files = sorted([f for f in os.listdir(path) if f.endswith(".png")])

# --- Load images ---
frames = [Image.open(os.path.join(path, f)) for f in png_files]

# --- Save as GIF ---
gif_path = os.path.join(path, "animation.gif")
frames[0].save(
    gif_path,
    save_all=True,
    append_images=frames[1:],
    duration=400,   # milliseconds per frame
    loop=0          # 0 = infinite loop
)
print(f"GIF saved to {gif_path}")

# --- Save as MP4 (easier to pause/rewind) ---
mp4_path = os.path.join(path, "animation.mp4")
images = [imageio.imread(os.path.join(path, f)) for f in png_files]
imageio.mimsave(mp4_path, images, fps=2.5)  # adjust fps (frames per second)
print(f"MP4 saved to {mp4_path}")
```
