```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "DATA/20140915H1.cat"

# --- Read in the .cat file ---
df = pd.read_fwf(
    file_path,
    skiprows=5,
    names=["Seq", "Serial", "DateTime", "Lat", "Lon", "Pr", "Comment1", "Comment2"],
    widths=[5, 12, 15, 8, 8, 8, 30, 30]
)

# --- Clean up ---
df = df.dropna(subset=["Lat", "Lon", "Pr"])  # remove blank rows

# Convert latitude, longitude, and pressure to numeric
df["Lat"] = pd.to_numeric(df["Lat"], errors="coerce")
df["Lon"] = pd.to_numeric(df["Lon"], errors="coerce")
df["Pr"]  = pd.to_numeric(df["Pr"], errors="coerce")

# Now safely flip sign for longitude
df["Lon"] = -df["Lon"]

# --- Combine comments ---
df["Comment"] = df["Comment1"].fillna("") + " " + df["Comment2"].fillna("")
df = df[["Seq", "Lat", "Lon", "Pr", "Comment"]]

print(df)

# --- Plot 1: Map of dropsonde positions ---
plt.figure(figsize=(7,6))
plt.scatter(df["Lon"], df["Lat"], c=df["Pr"], cmap="viridis", s=80, edgecolor='k')
for i, txt in enumerate(df["Seq"]):
    plt.text(df["Lon"].iloc[i]+0.05, df["Lat"].iloc[i]+0.05, str(txt), fontsize=8)
plt.title("Hurricane Edouard (2014) – Dropsonde Locations (2014-09-15H1)")
plt.xlabel("Longitude (°W)")
plt.ylabel("Latitude (°N)")
plt.colorbar(label="Surface Pressure (mb)")
plt.grid(True)
plt.show()

# --- Plot 2: Pressure vs Latitude ---
plt.figure(figsize=(6,5))
plt.scatter(df["Lat"], df["Pr"], c=df["Lon"], cmap="coolwarm", s=80, edgecolor='k')
plt.title("Pressure vs Latitude – Hurricane Edouard 2014 (20140915H1)")
plt.xlabel("Latitude (°N)")
plt.ylabel("Pressure (mb)")
plt.gca().invert_yaxis()
plt.colorbar(label="Longitude (°W)")
plt.grid(True)
plt.show()
```
